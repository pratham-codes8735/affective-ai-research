```python
import pandas as pd

data = {

    'text': [
        'I am so happy',
        'This is absolutely terrible',
        'I feel really nervous about tomorrow',
        'What a wonderful surprise',
        'I am so angry right now',
        ],
    'emotion':['joy', 'anger', 'fear','surprise','anger']
}

# pd.DataFrame converts the dictionary above into a table

df= pd.DataFrame(data)
print(df) 
```

```python
# How many examples of each emotion do we have right now ???  This is called checking of class distribution
# and i feel it is somewhat similar to statistics
print(df['emotion'].value_counts())
```

```python
# how long is each text ??(number of words)
#.apply() runs the function on every row 
df['word_count']=df['text'].apply(lambda x: len(x.split()))
print(df)
```

```python
angry_texts=df[df['emotion']== 'anger']
print(angry_texts)
```
