```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    'text':[
        'I am so happy today',
        'This is absolutely terrible',
        'I feel really nervous for tomorrow',
        'What a wonderful surprise',
        'I am so angry right now'
    ],
    'emotion':['joy','anger','fear','surprise','anger']
}

df=pd.DataFrame(data)
print(df)
```

```python
df['word_count'] = df['text'].apply(lambda x: len(x.split()))
word_counts_array= df['word_count'].values
print("As pandas series : ",df["word_count"])
print("As numpy array : ", word_counts_array)
print("Type : ",type(word_counts_array))
print("Mean word count : ", np.mean(word_counts_array))
```

```python
emotions = df['emotion'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(emotions.index,emotions.values,color='steelblue')
plt.title('Emotion Distribution of our mini data set')
plt.xlabel('Emotion')
plt.ylabel('Number of examples')
plt.show()
```

```python
# One more chart — word count per emotion
# Does text length differ by emotion? Let's visualise it
df['word_count'] = df['text'].apply(lambda x: len(x.split()))
plt.figure(figsize=(8, 5))
plt.bar(df['emotion'], df['word_count'], color='coral')
plt.title('Word count per text')
plt.xlabel('Emotion')
plt.ylabel('Word count')
plt.show()
```

```python


print("Summary of our dataset : ")
print(f"Total examples : {len(df)}")
print(f"Unique emotions : {df['emotion'].nunique()}")
print(f"Average word count : {np.mean(df['word_count'].values):.2f}")
print(f"Longest text : {df.loc[df['word_count'].idxmax(),'text']}")
print(f"Shortest text : {df.loc[df['word_count'].idxmin(),'text']}")
```
