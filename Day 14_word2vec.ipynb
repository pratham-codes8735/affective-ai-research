```python
from datasets import load_dataset
import pandas as pd
from gensim.models import Word2Vec 
import numpy as np

#load go emotions
dataset = load_dataset("google-research-datasets/go_emotions","simplified")
train_df=pd.DataFrame(dataset['train'])

#get all texts
all_texts = train_df['text'].tolist()
print(f"Total texts : {len(all_texts)}")
print(f"Example : {all_texts[0]}")
```

```python
'''
Word2Vec accepts tokenized words that is it accepts list of words , not full strings/sentences so we split each sentence into individual words
.lower() converts to lowercase so "Happy" and "happy" are same word
.split() splits on spaces

'''
tokenized_texts = [text.lower().split() for text in all_texts]

print(f"Example tokenized : {tokenized_texts[0]}")
print(f"type : {type(tokenized_texts[0])}")
```

```python
'''
"Why do we convert to lowercase before training Word2Vec? What would happen if we didn't?"
--> if we do not convert to lowercase then it will unnecessarily have two words which have exactly
the same meaning for example 'Happy' and'happy' have the same meaning and can be used multiple times
so instead of counting them as single , Word2Vec will have to store two words which is just the waste of space


'''
```

```python
'''
Train Word2Vec on GoEmotions Texts
vector_size = 100 : each word gets represented by 100 numbers
window=5: look at 5 words on each side to understand context
min_count=2 : ignore words that appear only once (typos)
workers = 4: use 4 CPU Ccores to speed up training
epochs=10 : pass through all texts 10 times

'''
model = Word2Vec(
    sentences = tokenized_texts,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    epochs=10    
)
print("Word2Vec model trained successfully!")
print(f"Vocabulary size : {len(model.wv)}")
```

```python
happy_vector=model.wv['happy']
print(f"Vector for 'happy' : shape = {happy_vector.shape}")
print(f"First 10 numbers : {happy_vector[:10]}")
```

```python
test_words=['happy','angry','sad','love','fear']
for word in test_words:
    if word in model.wv:
        similar = model.wv.most_similar(word,topn=5)
        print(f"\n Most similar to '{word}':")
        for similar_word,score in similar:
            print(f"{similar_word}:{score : .3f}")
    else:
        print(f" '{word}' not in vocabulary")
```

```python
try:
    result=model.wv.most_similar(
        positive=['joy','negative'],
        negative=['positive'],
        topn=3
    )
    print("joy-positive+negative=")
    for word , score in result:
        print (f" {word}:{score:.3f}")
except KeyError as e :
    print(f"Word not in vocabulary : {e}")
    
    
    
```
