```python
import numpy as np
#numpy is simply a combination of numbers and python 
#numpy array is just like a python list but built for mathematics

scores=np.array([0.9,0.3,0.7,0.1,0.8])
print ("Array : ", scores)
print("Type : ", type(scores))
```

```python
#Basic statistics on an array in numpy
print("Mean : ", np.mean(scores))
print("Max : ", np.max(scores))
print("Min : ", np.min(scores))
print("Standard deviation : ", np.std(scores))
```

```python
#if we want to create some filters like which scores are above 0.5
mask = scores > 0.5 
print("Mask : ", mask) # shows true for greater than 0.5 and false for less than 0.5
print("Filtered : ", scores[mask]) # shows only the true values
```

```python
#why we are learning and practicing all these???
#this is because when any of my models predict emotions , it returns array like the one we have created :
# [joy=0.72, sadness=0.10, anger=0.08, fear=0.06, surprise=0.04]
#we need numpy to find statistics of different emotions and not onlt this but i think that this same method is used in different cases 

emotion_labels=['joy','sadness','anger','fear','surprise']
model_output=np.array([0.72,0.10,0.08,0.06,0.04])

#np.argmax returns the index of the highest value 

predicted_index = np.argmax(model_output)
predicted_emotion = emotion_labels[predicted_index]

print("Predicted_emotion : " , predicted_emotion)
print("Confidience : ", model_output[predicted_index])
```
